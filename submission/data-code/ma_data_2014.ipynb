```python
base_2014 = pd.merge(
    plan_data_2014,
    service_data_2014[["contractid", "month", "year", "ssa", "fips", "county", "state", "partial", "notes"]],
    on=["contractid", "month", "year"],
    how="left",
    suffixes=("", "_service")
)

print(base_2014.shape)
base_2014.head()
```

```python
# Fill missing county information from service-area fields
base_2014["state"] = base_2014["state"].fillna(base_2014["state_service"])
base_2014["county"] = base_2014["county"].fillna(base_2014["county_service"])
base_2014["fips"] = base_2014["fips"].fillna(base_2014["fips_service"])
base_2014["ssa"] = base_2014["ssa"].fillna(base_2014["ssa_service"])
```

```python
base_2014[["contractid", "planid", "state", "county", "fips", "ssa", "enrollment", "plan_type", "partd", "snp"]].head()
```

```python
base_2014[["state", "county", "fips", "ssa"]].isna().sum()
```

```python
# Remove SNPs
base_2014 = base_2014[base_2014["snp"] != "Yes"]
print(base_2014.shape)
```

```python
# Remove 800-series plans
base_2014 = base_2014[(base_2014["planid"] < 800) | (base_2014["planid"] >= 900)]
print(base_2014.shape)
```

```python
# Remove prescription drug only plans
base_2014 = base_2014[base_2014["plan_type"] != "Prescription Drug Plan"]
print(base_2014.shape)
```

```python
base_2014.to_csv(output_path + "/base_data_2014.csv", index=False)
print("Saved base_data_2014.csv")
```
